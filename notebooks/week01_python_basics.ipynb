```python
# 各类基础数据类型
my_str = "Python基础学习"       # 字符串
my_int = 2026                  # 整数
my_float = 3.14159             # 浮点数
my_list = [10, 20, "hello", True]  # 列表
my_dict = {"name": "student", "grade": 1, "score": 95.5}  # 字典

# 打印查看
print("字符串：", my_str, type(my_str))
print("整数：", my_int, type(my_int))
print("浮点数：", my_float, type(my_float))
print("列表：", my_list, type(my_list))
print("字典：", my_dict, type(my_dict))
```

```python
def count_words(text):
    """统计英文句子单词数量，按空格分割"""
    word_list = text.strip().split()
    return len(word_list)

# 测试示例
test_sentence = "I love learning Python and GitHub"
print(f"句子：{test_sentence}")
print(f"单词总数：{count_words(test_sentence)}")
```
